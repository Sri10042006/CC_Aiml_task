20.Create a Box Plot for Sales. Are there any possible outliers?

```python

plt.boxplot(df["Sales"],vert=True)
plt.title("Sales Boxplot")
plt.ylabel("Sales")
plt.show()
```

21.Create a Bar Chart showing sales by Product_Category. Which category has the highest sales?


```python

plt.hist(df["Sales"])
count=df.groupby("Product_Category")["Sales"].mean()
count.plot(kind="bar")
plt.ylabel("Sales")
plt.xlabel("Product_Category")
plt.title("Total Sales by product category")
plt.show()

```
