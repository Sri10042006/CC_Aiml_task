Correlation

```python
import seaborn as sns

# Correlation of numerical variables
corr = df[["Quantity","Unit_Price","Discount","Delivery_Days","Sales"]].corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

Strongest Relatoinship:Delivery_days and unit price

Relationship:positive

Regression Model

```python
from sklearn.linear_model import LinearRegression
X = df[["Customer_Age", "Quantity", "Unit_Price", "Discount", "Delivery_Days"]]
y = df["Sales"]
model = LinearRegression()
model.fit(X, y)
#input features
X = df[["Customer_Age", "Quantity", "Unit_Price", "Discount", "Delivery_Days"]]
#target
y = df["Sales"]
#new order
new_order = pd.DataFrame({
    "Customer_Age": [30],
    "Quantity": [3],
    "Unit_Price": [1000],
    "Discount": [10],
    "Delivery_Days": [4]
})

print(new_order)
#Predict
predicted_sales = model.predict(new_order)
print(f"Predicted Sales: {predicted_sales[0]:.2f}")
```
